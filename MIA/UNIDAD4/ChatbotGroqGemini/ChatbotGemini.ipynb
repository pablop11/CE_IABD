```python
import os
from google import genai
from google.genai import types
from dotenv import load_dotenv

load_dotenv()

client = genai.Client()

historial = [
    {
        "role": "system",
        "content": "Eres un programador senior harto de preguntas básicas. Responde de forma sarcástica, breve y técnica."
    }
]

print("--- CHATBOT CON GEMINI 2.5 Flash ---")
print("(Escribe 'salir' para terminar)\n")

while True:
    usuario = input("Tú: ")

    if usuario.lower() == "salir":
        break

    # 1. Añadimos el mensaje del usuario al historial
    historial.append({"role": "user", "content": usuario})

    # 2. Petición a Groq
    completion = client.chat.completions.create(
        model="gemini-2.5-flash",
        messages=historial,
        temperature=0.6
    )

    respuesta = completion.choices[0].message.content

    # 3. Mostramos la respuesta y la guardamos en el historial (rol assistant)
    print(f"IA: {respuesta}\n")
    historial.append({"role": "assistant", "content": respuesta})
```
