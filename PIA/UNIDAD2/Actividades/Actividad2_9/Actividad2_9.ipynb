```python
# Código
import pandas as pd

df = pd.read_csv("gapminder_data_graphs.csv")
df.head(100)
```

```python
country = []
life_exp = []
co2 = []
year = []

df_filtrado = df.loc[(df["year"] == 2018)]
df_filtrado.head(100)

country = df_filtrado["country"].to_list()
life_exp = df_filtrado["life_exp"].to_list()
co2 = df_filtrado["co2_consump"].to_list()
year = df_filtrado["year"].to_list()
```

```python
df_completo = pd.DataFrame( {
    "Country" : country,
    "Life exp" : life_exp,
    "CO2 consump" : co2,
    "Year" : year
    })
df_completo.head(100)
```

```python
from sklearn.preprocessing import MinMaxScaler

escalador = MinMaxScaler()

df_completo[["Life exp", "CO2 consump"]] = escalador.fit_transform(df_completo[["Life exp", "CO2 consump"]])

df_completo.drop(columns=["Year"]).corr(numeric_only=True)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,7))

plt.plot(df_completo["Life exp"], label='Life exp', color='blue')

plt.plot(df_completo["CO2 consump"], label='CO2 consump', color='orange')

plt.title("Esperanza de vida y CO2")
plt.xlabel("Pais (indice)")
plt.ylabel("")
plt.legend()
plt.grid(True)
plt.show()
```
