# Análisis de Datos con el Dataset "Adult Income"

El Dataset "Adult Income" es una colección de datos demográficos y laborales extraídos del censo estadounidense de 1994, recopilado por la Oficina del Censo de los Estados Unidos. Este conjunto de datos es ampliamente utilizado en tareas de aprendizaje automático, estadística y análisis de datos para estudios relacionados con clasificación, regresión y pruebas de hipótesis estadísticas.

El objetivo principal asociado con este dataset es predecir si un individuo gana más de $50,000 al año en función de sus atributos personales y laborales. Este análisis permite explorar relaciones entre variables y realizar interpretaciones significativas a nivel social y económico.

## Descripción de las Variables

El dataset incluye las siguientes variables:

- **age**: Edad del individuo (en años).
- **workclass**: Tipo de empleo (p. ej., privado, gubernamental, autónomo, etc.).
- **fnlwgt**: Peso final asignado a la observación, utilizado para expandir la muestra del censo.
- **education**: Nivel educativo alcanzado (p. ej., secundaria, licenciatura, maestría).
- **education.num**: Nivel educativo en formato numérico (años de educación).
- **marital.status**: Estado civil (p. ej., casado, soltero, divorciado).
- **occupation**: Ocupación laboral (p. ej., administrativo, técnico, etc.).
- **relationship**: Relación familiar con el cabeza de familia (p. ej., esposo/a, hijo/a, etc.).
- **race**: Raza del individuo (p. ej., blanca, negra, asiática).
- **sex**: Género del individuo (masculino o femenino).
- **capital.gain**: Ganancia de capital obtenida.
- **capital.loss**: Pérdida de capital reportada.
- **native.country**: País de origen.
- **hours.per.week**: Número de horas trabajadas por semana.
- **income**: Variable objetivo, indica si el ingreso anual es '>50K' o '<=50K'.

```python
import pandas as pd

df_adult = pd.read_csv("adult.csv")
df_adult.head()
```

```python
df_adult_sin_capital_gain = df_adult.loc[df_adult["capital.gain"] > 0]
print(df_adult_sin_capital_gain["capital.gain"])
```

```python
'''1. Realiza un análisis general de los datos. Presta atención al tamaño de los datos y la cantidad de datos nulos que tiene el set de datos. Analiza y clasifica las variables del dataset.
- Indica el tamaño del dataset (número de filas y columnas).
- Comenta brevemente si la calidad de los datos es adecuada para el análisis posterior o si sería necesario algún tratamiento previo.

Selecciona las variables más apropiadas para realizar los siguientes cálculos para que aporten información relevante sobre los datos:
- Calcular medidas de tendencia central (media, mediana, moda) y dispersión (desviación estándar, varianza).
- Visualizar los datos mediante histogramas, diagramas de barras, diagramas de caja y gráficos de sectores.
- Detectar valores atípicos en las variables numéricas.
- Explorar relaciones entre variables numéricas y categóricas.
- Evaluar las distribuciones de las variables clave.'''

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

df_adult.info()
#df_adult.shape()

edad = df_adult["age"]

# Medidas de tendencia central
print("Media:", edad.mean())
print("Mediana:", edad.median())
print("Moda:", edad.mode().iloc[0])

# Medidas de dispersión
print("Desviación estándar:", edad.std())
print("Varianza:", edad.var())

# Histograma
plt.figure(figsize=(10, 6))
sns.histplot(x="age", y="capital.gain", data=df_adult_sin_capital_gain)
plt.title("Relación entre Edad y Capital ganado")
plt.xlabel("Edad")
plt.ylabel("Capital")
plt.show()

# Diagrama de barras
plt.figure(figsize=(10, 6))
sns.countplot(x="age", data=df_adult)
plt.xticks(ticks=range(0, 100, 4))
plt.title("Edad")
plt.xlabel("Edad")
plt.show()

# Diagrama de cajas
plt.figure(figsize=(10, 6))
sns.scatterplot(x="age", y="capital.gain", data=df_adult_sin_capital_gain)
plt.title("Relación entre Edad y Capital ganado")
plt.xlabel("Edad")
plt.ylabel("Capital")
plt.show()

# Diagrama de sectores
df_adult["age"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribución por edad")
plt.ylabel("")  
plt.show()

# Detectar outliers
outliersCapital = []
q25 = np.percentile(df_adult_sin_capital_gain["capital.gain"],[25])
q50 = np.percentile(df_adult_sin_capital_gain["capital.gain"],[50]) 
q75 = np.percentile(df_adult_sin_capital_gain["capital.gain"],[75]) 
iqrCapital = q75-q25

print("Cuartiles:", np.percentile(df_adult_sin_capital_gain["capital.gain"],[25, 50, 75]))
print("IQR:", iqrCapital)
print("Detectar outliers:", "menor que",q25-1.5*iqrCapital, "mayor que", q75+1.5*iqrCapital)

for c in df_adult_sin_capital_gain["capital.gain"].to_list():
    if float(c) < q25-1.5*iqrCapital or float(c) > q75+1.5*iqrCapital:
        outliersCapital.append(c)

print("Outliers:", outliersCapital)
```

```python
gana_mas_de_50K = []
for index,row in df_adult.iterrows():
    if row['income'] == ">50K":
        gana_mas_de_50K .append(True)
    else:
        gana_mas_de_50K .append(False)
df_adult['gana_mas_de_50k'] = gana_mas_de_50K 
df_adult.head(10)

conteo = df_adult.groupby(['sex', 'gana_mas_de_50k']).size().unstack(fill_value=0)

conteo.plot(kind='bar', figsize=(10, 6))
plt.title('Distribución de Ingresos por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Cantidad de Personas')
plt.legend(['Gana ≤50K', 'Gana >50K'], title='Ingresos')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
'''4. Consultas a los datos
- ¿Cuál es la media de la edad por nivel educativo?
- ¿Cuántos individuos trabajan más de 40 horas por semana y cuál es su distribución por ocupación?
- ¿Cuál es el promedio de ganancia de capital entre los individuos que tienen un ingreso mayor a $50,000?'''

niveles_educativos = []
medias = []

for index, row in df_adult.iterrows():
    if row["education.num"] not in niveles_educativos:
        niveles_educativos.append(row["education.num"])

        df_filtrado = df_adult.loc[df_adult["education.num"] == row["education.num"]]

        medias.append(np.mean(df_filtrado["age"]))

df = pd.DataFrame({
    'nivel_educativo': niveles_educativos,
    'media': medias
})

print(df.head(30))
#print(niveles_educativos)
#print(medias)

individuos = 0
for index, row in df_adult.iterrows():
    if row["hours.per.week"] > 40:
        individuos += 1
print("Individuos que trabajan mas de 40h semanales: ", individuos)

lista = []
media_ganancia_mas50k = 0
for index, row in df_adult.iterrows():
    if row["income"] == ">50K":
        lista.append(row["capital.gain"])
    
media_ganancia_mas50k = np.mean(lista)
#print("Lista: ", lista)
print("Promedio ganancia capital: ", media_ganancia_mas50k)

```
