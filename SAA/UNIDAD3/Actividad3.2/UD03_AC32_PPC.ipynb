# Descenso del gradiente en 3D

Este notebook muestra un ejemplo sencillo de descenso del gradiente en una función de dos variables:

$$f(x,y)=x^2+y^2$$

La idea es visualizar cómo el algoritmo desciende por una superficie hasta acercarse al mínimo.

## 1. Importar librerías

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

## 2. Definición de la función y de su gradiente

Trabajamos con la función:

$$f(x,y)=x^2+y^2$$

Su gradiente es:

$$\nabla f(x,y)=(2x,2y)$$

>Como hemos visto, el gradiente es la derivada (en este caso nos la dan ya calculada)

```python
def f(x, y):
    return x**2 + y**2

def grad(x, y):
    return 2*x, 2*y
```

## 3. Aplicación del Descenso del Gradiente

Partimos de un punto inicial y actualizamos los parámetros con la regla:
<!-- A continuación aparecen las expresiones. los símbolos son la forma en que Markdown nos permite manejar expresiones matemáticas -->
$$x_{nuevo}=x-\alpha \frac{\partial f}{\partial x}$$
$$y_{nuevo}=y-\alpha \frac{\partial f}{\partial y}$$

```python
# parámetros iniciales
alpha = 0.1
x, y = 3.5, 3.0
iteraciones = 15

# guardar trayectoria
xs = [x]
ys = [y]
zs = [f(x, y)]

# descenso del gradiente
for i in range(iteraciones):
    gx, gy = grad(x, y)
    x = x - alpha * gx
    y = y - alpha * gy
    xs.append(x)
    ys.append(y)
    zs.append(f(x, y))

for i in range(len(xs)):
    print(f"Iteración {i}: x = {xs[i]:.4f}, y = {ys[i]:.4f}, f(x,y) = {zs[i]:.4f}")
```

## 4. Representación de la superficie y de la trayectoria

```python
# malla para dibujar la superficie
x_vals = np.linspace(-4, 4, 100)
y_vals = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = f(X, Y)

# gráfico 3D
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, alpha=0.6)
ax.plot(xs, ys, zs, marker='o')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x,y)')
ax.set_title('Descenso del gradiente en 3D')

plt.show()
```

```python
plt.contour(X, Y, Z, levels=30)
plt.plot(xs, ys, 'ro-')
plt.title("Trayectoria del descenso de gradiente")
plt.show()
```

```python
# tres alphas distintos para el paraboloide
def run(alpha, x0, y0, iters=20):
    x, y = x0, y0
    traj = [(x,y)]
    for _ in range(iters):
        gx, gy = grad(x,y)
        x -= alpha * gx
        y -= alpha * gy
        traj.append((x,y))
    return np.array(traj)

# muy pequeño, razonable, demasiado grande
alphas = [0.001, 0.1, 1.0]          
colors = ['r','g','b']
x0, y0 = 3.0, -2.0
iters = 30

plt.contour(X, Y, Z, levels=30)
for a,c in zip(alphas, colors):
    traj = run(a, x0, y0, iters)
    plt.plot(traj[:,0], traj[:,1], marker='o', color=c,
             label=f'α={a}')
plt.legend()
plt.title('Tres trayectorias con distintos α (paraboloide)')
plt.xlabel('x'); plt.ylabel('y')
plt.show()
```

## 6.1 Tipo Silla

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Crear malla
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

# Superficie tipo "silla"
Z = X**2 - Y**2

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, alpha=0.8)

ax.set_title("Superficie tipo saddle: f(x,y) = x² - y²")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")

plt.show()
```

## 6.3. Función de Rosenbrock

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Crear malla
x = np.linspace(-2, 2, 200)
y = np.linspace(-1, 3, 200)
X, Y = np.meshgrid(x, y)

# Función de Rosenbrock
Z = (1 - X)**2 + 100*(Y - X**2)**2

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap="viridis")

ax.set_title("Función de Rosenbrock")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")

plt.show()
```

## 7. Cuestiones
Contesta con tus palabras las siguientes cuestiones sobre los tres casos resueltos (celdas 5, 6.1. y 6.2.):
1. ¿En cuál crees que el descenso de gradiente converge más fácilmente?
    -   En el paraboloide f(x,y) = x² + y². Al ser una función convexa y simétrica, tiene un único mínimo global en (0,0) y su gradiente apunta siempre de forma suave hacia él. El algoritmo desciende de manera estable en cada iteración sin riesgo de quedar atrapado.

2. ¿En cuál puede tener más dificultad?
    -   En la función de Rosenbrock f(x,y) = (1−x)² + 100(y−x²)². Presenta un valle muy estrecho y curvo que hace que el algoritmo avance muy lentamente a lo largo de él. Además, el gradiente tiene magnitudes muy distintas en cada dirección, lo que dificulta elegir un alpha adecuado.

3. ¿Por qué la función de Rosenbrock se utiliza para probar algoritmos de optimización?
    -   Precisamente porque es difícil de optimizar: su mínimo global está en (1,1) pero se encuentra al fondo de un valle estrecho y muy curvado. Un buen algoritmo de optimización debe ser capaz de navegar por ese valle sin divergir ni converger demasiado lento, lo que la convierte en un banco de pruebas exigente y representativo de problemas reales. 

4. ¿Por qué el descenso de gradiente sigue siempre la dirección del gradiente negativo?
    -   El gradiente indica la dirección en la que la función crece más rápidamente. Para minimizar la función hay que moverse en la dirección contraria, es decir, restando el gradiente en cada actualización.

5. ¿En cuál de las tres superficies el descenso de gradiente debería converger más rápidamente? 
    -   En el paraboloide, ya que su geometría es regular y simétrica, los gradientes son proporcionales a la distancia al mínimo y el descenso es directo y sin obstáculos en cada iteración.

6. ¿En cuál podría tardar más o presentar oscilaciones?  
    -   En la Rosenbrock, donde el paso puede provocar oscilaciones, la silla tambien puede divergir en la direccion concava si el paso es grande.

7. ¿Por qué crees que en problemas reales de Machine Learning la superficie de error puede ser mucho más compleja?
    -   En ML real hay millones de parámetros; la función de coste es no convexa, con mínimos locales, mesetas, puntos de silla y ruido, por lo que el paisaje es mucho más complejo.
