```python
'''2. Medidas de tendencia central
Con tiempos = [1.2, 1.5, 1.7, 2.0, 1.3, 1.4, 1.6, 2.1]:
- Calcula media, mediana y moda.
- Indica si existe moda real.
No hay una moda real'''

import numpy as np
from scipy import stats

datos = [1.2, 1.5, 1.7, 2.0, 1.3, 1.4, 1.6, 2.1]

print("Media:", np.mean(datos))
print("Mediana:", np.median(datos))
print("Moda:", stats.mode(datos)[0])

```

```python
'''3. Cuartiles e IQR
Con visualizaciones = [120, 130, 150, 160, 200, 210, 300, 320]:
- Calcula Q1, Q2, Q3.
- Calcula el IQR.
- Detecta outliers.
- Representa un boxplot.'''

import matplotlib.pyplot as plt

datos = [120, 130, 150, 160, 200, 210, 300, 320]
outliers = []

q1 = np.percentile(datos,[25]) 
q2 = np.percentile(datos,[50]) 
q3 = np.percentile(datos,[75]) 
iqr = q3-q1

print("Cuartiles:", np.percentile(datos,[25, 50, 75]))
print("IQR:", iqr)
print("Detectar outliers:", "menor que", q1-1.5*iqr, "mayor que", q3+1.5*iqr)

for d in datos:
    if float(d) < q1-1.5*iqr or float(d) > q3+1.5*iqr:
        outliers.append(d)

print("Outliers:", outliers)

plt.boxplot(datos)
plt.show()
```

```python
'''4. Comparación A/B
Sean
A = [30, 35, 50, 55, 40, 45, 60] B = [20, 25, 30, 35, 28, 32, 34]
- Media, desviación típica y CV.
- ¿Qué grupo es más constante?
- Representa gráficamente.'''

listaA = [30, 35, 50, 55, 40, 45, 60] 
listaB = [20, 25, 30, 35, 28, 32, 34]

mediaA = np.mean(listaA)
mediaB = np.mean(listaB)

desviacionA = np.std(listaA)
desviacionB = np.std(listaB)

coeficienteA = (desviacionA/mediaA)*100
coeficienteB = (desviacionB/mediaB)*100

print("Media A:", mediaA, " Media B:", mediaB)
print("Desviacion A:", desviacionA, " Desviacion B:", np.std(listaB))
print("Coeficiente de varianza A:", coeficienteA, " Coeficiente de varianza B:", coeficienteB)

plt.plot(listaA)
plt.plot(listaB)
plt.show()
```

```python
'''5. Histograma y asimetría
Con tiempos = [100, 102, 105, 110, 115, 120, 200, 220, 250, 300]:
- Histograma.
- Asimetría.
- Interpretación.
'''

import pandas as pd

datos = [100, 102, 105, 110, 115, 120, 200, 220, 250, 300]

np_array = pd.Series(datos)
asimetria = np_array.skew()

print("Asimetria:", asimetria)

plt.hist(datos)
plt.show()
```

```python
'''6. Estudio de Outliers
Con edades = [25, 26, 27, 25, 28, 29, 120]:
- Detecta outliers.
- Elimínalos.
- Encuentra y justifica un valor adecuado para sustituir los outliers.'''

edades = [25, 26, 27, 25, 28, 29, 120]
outliersEdad = []

q25 = np.percentile(edades,[25]) 
q50 = np.percentile(edades,[50]) 
q75 = np.percentile(edades,[75]) 
iqrEdad = q75-q25

print("Cuartiles:", np.percentile(edades,[25, 50, 75]))
print("IQR:", iqrEdad)
print("Detectar outliers:", "menor que",q25-1.5*iqrEdad, "mayor que", q75+1.5*iqrEdad)

for e in edades:
    if float(e) < q25-1.5*iqrEdad or float(e) > q75+1.5*iqrEdad:
        outliersEdad.append(e)

print("Outliers:", outliersEdad)
```

```python
'''7. Probabilidad condicionada
Dadas las siguientes probabilidades
P(móvil)=0.40, P(tablet)=0.30, P(PC)=0.30. P(corto|móvil)=0.80, P(corto|tablet)=0.50, P(corto|PC)=0.20.
- Calcula P(corto) y P(móvil|corto).'''

pmovil = 0.40
ptablet = 0.30
ppc = 0.30

pcortomovil = 0.80
pcortotablet = 0.50
pcortopc = 0.20

pcorto = (pcortomovil*pmovil) + (pcortotablet*pcortotablet) + (pcortopc*ppc)
pmovilcorto = pcortomovil*pmovil/pcorto

print("P(Corto):", pcorto)
print("P(Movil|Corto):", pmovilcorto)
```

```python
'''8. Teorema de Bayes
Dadas las siguientes probalidades
P(fraude)=0.02, P(alarma|fraude)=0.95, P(alarma|no fraude)=0.05.
- Calcula P(fraude|alarma).'''

pfraude = 0.02
pnofraude = 0.98

palarmafraude = 0.95
palarmanofraude = 0.05

palarma = palarmafraude*pfraude + palarmanofraude*pnofraude

pfraudealarma = palarmafraude*pfraude / palarma

print("P(Fraude|Alarma):", pfraudealarma)
```

```python
'''9. Regresión lineal
Dadas las siguientes variables y sus correspondientes valores, Publicidad: [1000, 1500, 2000, 2500]
Ventas: [120, 150, 170, 200]
- Ajusta una regresión lineal.
- Predice para 3000€.'''

import numpy as np
from sklearn.linear_model import LinearRegression

publicidad = np.array([1000, 1500, 2000, 2500]).reshape(-1, 1) 
ventas = np.array([120, 150, 170, 200])  

modelo = LinearRegression()

# Ajustar el modelo con los datos
modelo.fit(publicidad, ventas)

# Predecir las ventas para 3000€ de publicidad ([0] para que nos de los decimales)
publicidad_nueva = np.array([[3000]])
prediccion_ventas = modelo.predict(publicidad_nueva)[0]

print("Predicción de ventas para 3000€ de publicidad:", prediccion_ventas)
```
