# EJEMPLO GRADIENTE

## 0. Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. ALgoritmo

>El descenso del gradiente busca el mínimo de una función moviéndose en la dirección opuesta al gradiente.

```python
def gradient_descent(f, grad, x0, alpha=0.1, iteraciones=20):
    
    x = x0
    xs = [x]

    for i in range(iteraciones):
        x = x - alpha * grad(x)
        xs.append(x)

    return xs
```

### 1.1 Definiciones

```python
# ---- FUNCIONES A OPTIMIZAR ----

# 1. Función cuadrática
def f_cuadratica(x):
    return x**2

def grad_cuadratica(x):
    return 2*x


# 2. Función con mínimo plano
def f_polinomio4(x):
    return x**4

def grad_polinomio4(x):
    return 4*x**3


# 3. Función con mínimos locales
def f_minimos_locales(x):
    return x**4 - 3*x**3 + 2

def grad_minimos_locales(x):
    return 4*x**3 - 9*x**2
```

## 2. Función cuadrática (Polinomio de grado dos)

```python


xs = gradient_descent(f_cuadratica, grad_cuadratica, x0=4, alpha=0.1)

x_vals = np.linspace(-5,5,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_cuadratica(x_vals), label="f(x)=x²")
plt.scatter(xs, f_cuadratica(np.array(xs)), color="red")
plt.plot(xs, f_cuadratica(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - función cuadrática")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-1, 17)
plt.xlim(-5, 5)
plt.show()
```

## 3. Polinomio de grado 4

```python


xs = gradient_descent(f_polinomio4, grad_polinomio4, x0=2, alpha=0.05)

x_vals = np.linspace(-3,3,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_polinomio4(x_vals), label="f(x)=x⁴")
plt.scatter(xs, f_polinomio4(np.array(xs)), color="red")
plt.plot(xs, f_polinomio4(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - convergencia lenta")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-1, 20)
plt.xlim(-3, 3)
plt.show()
```

## 4. Función con mínimos locales

```python


xs = gradient_descent(f_minimos_locales, grad_minimos_locales, x0=0.5, alpha=0.01)

x_vals = np.linspace(-1,3,200)

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_minimos_locales(x_vals), label="f(x)=x⁴-3x³+2")
plt.scatter(xs, f_minimos_locales(np.array(xs)), color="red")
plt.plot(xs, f_minimos_locales(np.array(xs)), linestyle="--")

plt.title("Descenso del gradiente - mínimos locales")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.ylim(-10, 10)
plt.xlim(-1, 3.5)
plt.show()
```

## 5. Comparamos los tres casos

```python
fig, axs = plt.subplots(1,3, figsize=(15,4))

# ---- CASO 1 ----
def f(x):
    return x**2

def grad(x):
    return 2*x

xs = gradient_descent(f, grad, x0=4, alpha=0.1)

x_vals = np.linspace(-5,5,200)

axs[0].plot(x_vals, f(x_vals))
axs[0].scatter(xs, f(np.array(xs)), color="red")
axs[0].plot(xs, f(np.array(xs)), linestyle="--")
axs[0].set_title("f(x)=x²")
axs[0].grid()


# ---- CASO 2 ----
def f(x):
    return x**4

def grad(x):
    return 4*x**3

xs = gradient_descent(f, grad, x0=2, alpha=0.05)

x_vals = np.linspace(-3,3,200)

axs[1].plot(x_vals, f(x_vals))
axs[1].scatter(xs, f(np.array(xs)), color="red")
axs[1].plot(xs, f(np.array(xs)), linestyle="--")
axs[1].set_title("f(x)=x⁴ (mínimo plano)")
axs[1].grid()


# ---- CASO 3 ----
def f(x):
    return x**4 - 3*x**3 + 2

def grad(x):
    return 4*x**3 - 9*x**2

xs = gradient_descent(f, grad, x0=0.5, alpha=0.01)

x_vals = np.linspace(-1,3,200)

axs[2].plot(x_vals, f(x_vals))
axs[2].scatter(xs, f(np.array(xs)), color="red")
axs[2].plot(xs, f(np.array(xs)), linestyle="--")
axs[2].set_title("mínimos locales")
axs[2].grid()


plt.suptitle("Comparación del descenso del gradiente en distintas funciones")

plt.show()
```

## 6. Completa el siguiente caso

```python
import numpy as np
import matplotlib.pyplot as plt

# dataset sencillo
x_data = np.array([1,2,3,4])
y_data = np.array([2,4,6,8])

def f(w):
    y_pred = w * x_data
    return np.mean((y_data - y_pred)**2)

def grad(w):
    y_pred = w * x_data
    return -2*np.mean(x_data*(y_data - y_pred))

# descenso del gradiente
w = 0
alpha = 0.01
iteraciones = 50
ws = [w]
for i in range(iteraciones):
    w = w - alpha * grad(w)
    ws.append(w)

# --- PLOTS ---
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Curva de error f(w) con trayectoria del gradiente
w_vals = np.linspace(0, 3, 300)
f_vals = [f(wi) for wi in w_vals]
error_trayectoria = [f(wi) for wi in ws]

axes[0].plot(w_vals, f_vals, 'b-', linewidth=2, label='f(w) = MSE')
axes[0].plot(ws, error_trayectoria, 'ro-', markersize=4, label='Trayectoria')
axes[0].plot(ws[0], error_trayectoria[0], 'gs', markersize=10, label='Inicio')
axes[0].plot(ws[-1], error_trayectoria[-1], 'r*', markersize=12, label='Final')
axes[0].set_xlabel('w')
axes[0].set_ylabel('MSE')
axes[0].set_title('Descenso del Gradiente sobre f(w)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. Evolución de w por iteración
axes[1].plot(range(len(ws)), ws, 'b-o', markersize=3)
axes[1].axhline(y=2, color='r', linestyle='--', label='w óptimo = 2')
axes[1].set_xlabel('Iteración')
axes[1].set_ylabel('w')
axes[1].set_title('Convergencia de w')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# 3. Ajuste final vs datos reales
x_line = np.linspace(0, 5, 100)
axes[2].scatter(x_data, y_data, color='red', s=100, zorder=5, label='Datos reales')
axes[2].plot(x_line, ws[-1] * x_line, 'b-', linewidth=2, label=f'y = {ws[-1]:.3f}·x')
axes[2].set_xlabel('x')
axes[2].set_ylabel('y')
axes[2].set_title('Recta ajustada')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
